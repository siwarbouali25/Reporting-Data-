# Emirates NBD IFRS S1/S2 Style Extraction — Azure OpenAI Debug Notebook

This version fixes the two main PDF extraction problems:

- **Concatenated words**: it reconstructs text from PDF word coordinates using `pdfplumber.extract_words()` instead of relying on raw page text.
- **Table of contents leakage**: it detects and skips pages that look like Contents/Table of Contents before building section chunks.

Output: `emirates_nbd_style_reference.json`

````python
# ── Section detection config ────────────────────────────────
# This version uses heading/range detection first.
# Keyword classification is kept only as a fallback/debug helper.

SECTION_ORDER = [
    "general",
    "governance",
    "strategy",
    "risk_management",
    "metrics_and_targets",
]

SECTION_DISPLAY_NAMES = {
    "general": "General Requirements",
    "governance": "Governance",
    "strategy": "Strategy",
    "risk_management": "Risk management",
    "metrics_and_targets": "Metrics and targets",
}

SECTION_HEADING_PATTERNS = {
    "general": [
        r"^\s*1\.?\s*General Requirements\s*$",
        r"^\s*General Requirements\s*$",
    ],
    "governance": [
        r"^\s*2\.?\s*Governance\s*$",
        r"^\s*Governance\s*$",
    ],
    "strategy": [
        r"^\s*3\.?\s*Strategy\s*$",
        r"^\s*Strategy\s*$",
    ],
    "risk_management": [
        r"^\s*4\.?\s*Risk management\s*$",
        r"^\s*Risk management\s*$",
    ],
    "metrics_and_targets": [
        r"^\s*5\.?\s*Metrics and targets\s*$",
        r"^\s*Metrics and Targets\s*$",
        r"^\s*Metrics & targets\s*$",
        r"^\s*Metrics & Targets\s*$",
    ],
}

# Fallback keyword classifier only.
SECTION_KEYWORDS = {
    "governance": [
        "governance", "board oversight", "board of directors",
        "management role", "committee", "oversight"
    ],
    "strategy": [
        "strategy", "strategic", "climate-related risks", "opportunities",
        "scenario analysis", "transition plan", "resilience",
        "time horizon", "physical risk", "transition risk"
    ],
    "risk_management": [
        "risk management", "risk identification", "risk assessment",
        "enterprise risk", "climate risk integration", "risk appetite",
        "risk framework"
    ],
    "metrics_and_targets": [
        "metrics", "targets", "scope 1", "scope 2", "scope 3",
        "ghg emissions", "carbon", "tco2e", "net zero",
        "financed emissions", "baseline year", "emissions intensity"
    ],
}

SKIP_PAGE_PATTERNS = [
    r"^table of contents",
    r"^contents$",
    r"^appendix",
    r"^\d+$",
]

MAX_CHUNK_WORDS = 1500

print("Section heading detection config ready")
````

````python
# ── Step 1: PDF page extraction and cleaning ────────────────
def clean_pdf_text(text: str) -> str:
    """
    Light text cleaner.
    Important: do NOT try aggressive word repair here.
    The real fix for concatenated words is extracting words by coordinates.
    """
    if not text:
        return ""

    text = text.replace("\x08", " ")
    text = text.replace("\u00a0", " ")

    cleaned_lines = []
    for line in text.splitlines():
        line = re.sub(r"\s+", " ", line).strip()
        if line:
            cleaned_lines.append(line)

    text = "\n".join(cleaned_lines)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()


def _normalise_word_text(word: str) -> str:
    """
    Normalise one extracted word/token.
    """
    word = str(word or "").strip()
    word = word.replace("\u00a0", " ")
    word = re.sub(r"\s+", " ", word).strip()
    return word


def _page_words_to_lines(words: list[dict], y_tolerance: float = 3.0, column_gap: float = 95.0) -> str:
    """
    Reconstruct page text from pdfplumber word boxes.

    Why:
    page.extract_text() can produce bad strings like:
    'I2024thGrouharefresheouexistindoublmaterialitassessment'

    extract_words() gives x/y coordinates for each word, so we rebuild lines and join words with spaces.
    """
    if not words:
        return ""

    items = []
    for w in words:
        txt = _normalise_word_text(w.get("text", ""))
        if not txt:
            continue
        items.append({
            "text": txt,
            "x0": float(w.get("x0", 0)),
            "x1": float(w.get("x1", 0)),
            "top": float(w.get("top", 0)),
            "bottom": float(w.get("bottom", 0)),
        })

    if not items:
        return ""

    items.sort(key=lambda x: (x["top"], x["x0"]))

    lines = []
    current = []
    current_top = None

    for item in items:
        if current_top is None:
            current = [item]
            current_top = item["top"]
            continue

        if abs(item["top"] - current_top) <= y_tolerance:
            current.append(item)
        else:
            lines.append(current)
            current = [item]
            current_top = item["top"]

    if current:
        lines.append(current)

    reconstructed_lines = []

    for line_items in lines:
        line_items.sort(key=lambda x: x["x0"])

        segments = []
        current_segment = []
        previous_x1 = None

        for item in line_items:
            if previous_x1 is not None and (item["x0"] - previous_x1) > column_gap:
                if current_segment:
                    segments.append(" ".join(current_segment))
                current_segment = [item["text"]]
            else:
                current_segment.append(item["text"])

            previous_x1 = max(previous_x1 or item["x1"], item["x1"])

        if current_segment:
            segments.append(" ".join(current_segment))

        reconstructed_lines.append("    ".join(segments))

    return clean_pdf_text("\n".join(reconstructed_lines))


def _extract_page_text_with_pdfplumber_words(pdf_path: Path, debug: bool = False) -> list[dict]:
    """
    Primary extraction method.
    Uses word coordinates to avoid concatenated text.
    """
    pages = []

    with pdfplumber.open(pdf_path) as pdf:
        total = len(pdf.pages)
        log.info(f"Total pages: {total}")

        for i, page in enumerate(tqdm(pdf.pages, desc="Extracting pages from word coordinates")):
            try:
                words = page.extract_words(
                    x_tolerance=2,
                    y_tolerance=3,
                    keep_blank_chars=False,
                    use_text_flow=False,
                    horizontal_ltr=True,
                    vertical_ttb=True,
                    extra_attrs=[],
                )
            except TypeError:
                words = page.extract_words(
                    x_tolerance=2,
                    y_tolerance=3,
                    keep_blank_chars=False,
                    use_text_flow=False,
                )

            text = _page_words_to_lines(words, y_tolerance=3.0, column_gap=95.0)
            text = clean_pdf_text(text)

            if not text or len(text) < 80:
                continue

            pages.append({
                "page_num": i + 1,
                "text": text,
                "word_count": len(text.split()),
                "extractor": "pdfplumber_extract_words_coordinates",
            })

    return pages


def _extract_page_text_with_pdfplumber_text(pdf_path: Path, debug: bool = False) -> list[dict]:
    """
    Fallback only. Raw extract_text can concatenate words, so this is not preferred.
    """
    pages = []

    with pdfplumber.open(pdf_path) as pdf:
        for i, page in enumerate(tqdm(pdf.pages, desc="Fallback extract_text")):
            text = (
                page.extract_text(
                    x_tolerance=2,
                    y_tolerance=3,
                    layout=True,
                    keep_blank_chars=False,
                    use_text_flow=False,
                )
                or ""
            )
            text = clean_pdf_text(text)

            if not text or len(text) < 80:
                continue

            pages.append({
                "page_num": i + 1,
                "text": text,
                "word_count": len(text.split()),
                "extractor": "pdfplumber_extract_text_fallback",
            })

    return pages


def looks_garbled_or_concatenated(text: str) -> bool:
    """
    Detect text that likely has concatenated words.
    """
    if not text:
        return True

    long_alpha_tokens = re.findall(r"\b[A-Za-z]{22,}\b", text)
    if len(long_alpha_tokens) >= 3:
        return True

    repeated_pairs = len(re.findall(r"([A-Za-z])\1{1,}", text))
    if repeated_pairs > 80:
        return True

    return False


def is_table_of_contents_page(text: str) -> bool:
    """
    Detect TOC/Contents pages so they do not become style chunks.
    """
    t = clean_pdf_text(text).lower()
    first_1800 = t[:1800]

    toc_words = [
        "contents",
        "table of contents",
        "general requirements",
        "governance",
        "strategy",
        "risk management",
        "metrics and targets",
        "materiality assessment",
    ]

    signal_count = sum(w in first_1800 for w in toc_words)
    has_many_page_number_lines = len(re.findall(r"\b\d{1,3}\b", first_1800)) >= 10

    if "contents" in first_1800 and signal_count >= 4:
        return True

    if signal_count >= 5 and has_many_page_number_lines:
        return True

    return False


def should_skip_page(text: str, debug: bool = False) -> bool:
    text = clean_pdf_text(text)
    if not text:
        return True

    first_line = text.split("\n")[0].lower().strip()

    if any(re.match(p, first_line) for p in SKIP_PAGE_PATTERNS):
        return True

    if is_table_of_contents_page(text):
        return True

    return False


def extract_pdf_pages(pdf_path: str | Path, debug: bool = False) -> list[dict]:
    """
    Extract pages with coordinate-based word reconstruction.
    """
    pdf_path = Path(pdf_path)
    log.info(f"Opening PDF: {pdf_path}")

    pages = _extract_page_text_with_pdfplumber_words(pdf_path, debug=debug)

    if not pages:
        log.warning("Word-coordinate extraction returned no pages. Falling back to extract_text.")
        pages = _extract_page_text_with_pdfplumber_text(pdf_path, debug=debug)

    before_skip = len(pages)
    pages = [p for p in pages if not should_skip_page(p["text"], debug=debug)]

    log.info(f"Extracted {len(pages)} content pages after TOC skipping ({before_skip - len(pages)} skipped)")
    if pages:
        log.info(f"Extractor used: {pages[0].get('extractor')}")

    bad_pages = [p["page_num"] for p in pages if looks_garbled_or_concatenated(p["text"])]
    if bad_pages:
        print("WARNING: possible concatenated/garbled text on pages:", bad_pages[:20])
        print("If these pages matter, inspect them manually or adjust x_tolerance/y_tolerance.")

    return pages


print("Coordinate-based PDF extraction ready")
````

````python
# ── Step 2: Heading-based section detection ────────────────
def classify_page_section(page_text: str) -> Optional[str]:
    """
    Fallback classifier based on keyword density.
    Heading detection should be used first for actual chunk building.
    """
    text_lower = clean_pdf_text(page_text).lower()
    scores = {}

    for section, keywords in SECTION_KEYWORDS.items():
        score = sum(text_lower.count(kw) for kw in keywords)
        if score > 0:
            scores[section] = score

    if not scores:
        return None

    return max(scores, key=scores.get)


def classify_page_scores(page_text: str) -> dict:
    """
    Debug helper: return all keyword scores.
    """
    text_lower = clean_pdf_text(page_text).lower()
    return {
        section: sum(text_lower.count(kw) for kw in keywords)
        for section, keywords in SECTION_KEYWORDS.items()
    }


def _line_matches_any_pattern(line: str, patterns: list[str]) -> bool:
    line = clean_pdf_text(line).strip()
    return any(re.match(pattern, line, flags=re.IGNORECASE) for pattern in patterns)


def detect_section_start_pages(pages: list[dict]) -> dict[str, int]:
    """
    Find likely start page for each major report section using headings near the top of content pages.
    Returns {section: page_num}.
    """
    starts = {}

    for page in pages:
        text = clean_pdf_text(page["text"])
        lines = [ln.strip() for ln in text.splitlines() if ln.strip()]
        first_lines = lines[:25]
        first_block = "\n".join(first_lines)

        # Skip pages that still look like TOC.
        if is_table_of_contents_page(text):
            continue

        for section, patterns in SECTION_HEADING_PATTERNS.items():
            if section in starts:
                continue

            # Direct line match near top.
            direct = any(_line_matches_any_pattern(line, patterns) for line in first_lines)

            # Sometimes heading is embedded with page header; allow very early block match.
            embedded = False
            for pattern in patterns:
                loose = pattern.replace("^", "").replace("$", "")
                if re.search(loose, first_block, flags=re.IGNORECASE):
                    # Avoid TOC-like matches with many other section names in same block.
                    other_section_mentions = sum(
                        SECTION_DISPLAY_NAMES[s].lower() in first_block.lower()
                        for s in SECTION_ORDER
                        if s != section
                    )
                    embedded = other_section_mentions <= 1

            if direct or embedded:
                starts[section] = page["page_num"]

    return starts


def group_pages_by_heading_ranges(pages: list[dict]) -> tuple[dict[str, list[dict]], dict[str, int]]:
    """
    Build section pages from heading start pages.
    This avoids the earlier bad behaviour where Contents/General pages were classified as Strategy/Governance.
    """
    starts = detect_section_start_pages(pages)

    sections = {s: [] for s in SECTION_ORDER}
    sections["unclassified"] = []

    if len(starts) < 3:
        log.warning("Not enough section headings detected. Falling back to keyword page classification.")
        for page in pages:
            section = classify_page_section(page["text"])
            if section:
                sections[section].append(page)
            else:
                sections["unclassified"].append(page)
        return sections, starts

    ordered_starts = sorted(starts.items(), key=lambda kv: kv[1])

    for idx, (section, start_page) in enumerate(ordered_starts):
        end_page = ordered_starts[idx + 1][1] if idx + 1 < len(ordered_starts) else 10**9
        sections[section] = [
            p for p in pages
            if start_page <= p["page_num"] < end_page
        ]

    assigned_pages = {p["page_num"] for plist in sections.values() for p in plist}
    sections["unclassified"] = [p for p in pages if p["page_num"] not in assigned_pages]

    return sections, starts


def group_pages_by_section(pages: list[dict]) -> dict[str, list[dict]]:
    """
    Main grouping function used by the notebook.
    Uses heading ranges first; keyword classification only as fallback.
    """
    sections, starts = group_pages_by_heading_ranges(pages)

    print("Detected section start pages:")
    for section in SECTION_ORDER:
        print(f"- {section}: {starts.get(section)}")

    print("\nSection page ranges/counts:")
    for name, pages_list in sections.items():
        if pages_list:
            page_nums = [p["page_num"] for p in pages_list]
            print(f"- {name}: {len(pages_list)} pages | {min(page_nums)}-{max(page_nums)}")
        else:
            print(f"- {name}: 0 pages")

    return sections


print("Heading-based section detection functions ready")
````

````python
# ── Step 3: Build section chunks ────────────────────────────
def trim_section_text_to_heading(section: str, text: str) -> str:
    """
    If a section chunk starts before the actual heading, trim to the first matching heading.
    """
    text = clean_pdf_text(text)
    patterns = SECTION_HEADING_PATTERNS.get(section, [])

    for pattern in patterns:
        match = re.search(pattern, text, flags=re.IGNORECASE | re.MULTILINE)
        if match:
            return text[match.start():].strip()

    return text


def remove_likely_headers_footers(text: str) -> str:
    """
    Remove recurring low-value page headers/footers from chunks.
    """
    cleaned = []
    for line in text.splitlines():
        l = line.strip()
        ll = l.lower()

        if not l:
            continue
        if re.fullmatch(r"\d+", l):
            continue
        if "emirates nbd" in ll and len(l.split()) <= 8:
            continue
        if "ifrs s1" in ll and "ifrs s2" in ll and len(l.split()) <= 12:
            continue
        if ll in {"contents", "general requirements", "governance", "strategy", "risk management", "metrics and targets"}:
            # keep actual section heading handled by trim; remove repeated headers
            cleaned.append(l)
            continue

        cleaned.append(l)

    return "\n".join(cleaned)


def build_section_chunk(pages: list[dict], section: str | None = None, max_words: int = MAX_CHUNK_WORDS) -> str:
    """
    Concatenate pages for a section up to max_words.
    Uses cleaned text and trims to the actual section heading when possible.
    """
    if not pages:
        return ""

    combined_text = "\n\n".join(clean_pdf_text(page["text"]) for page in pages)
    combined_text = remove_likely_headers_footers(combined_text)

    if section:
        combined_text = trim_section_text_to_heading(section, combined_text)

    words = combined_text.split()
    if len(words) > max_words:
        combined_text = " ".join(words[:max_words])

    return combined_text.strip()


print("build_section_chunk() ready")
````

````python
# ── Run Step 3: Build chunks ────────────────────────────────
section_chunks = {}

# Analyze only the real report sections. Do not send unclassified/TOC-like content to the style model.
sections_to_analyze = [
    "general",
    "governance",
    "strategy",
    "risk_management",
    "metrics_and_targets",
]

for section in sections_to_analyze:
    pg_list = section_pages.get(section, [])
    if pg_list:
        chunk = build_section_chunk(pg_list, section=section, max_words=MAX_CHUNK_WORDS)
        section_chunks[section] = chunk
        print(f"{section}: {len(chunk.split())} words | pages {[p['page_num'] for p in pg_list[:3]]}{'...' if len(pg_list) > 3 else ''}")
    else:
        print(f"{section}: 0 pages -> no chunk")

print("\nChunk preview:")
for section, chunk in section_chunks.items():
    print("\n" + "="*80)
    print(section.upper())
    print("="*80)
    print(chunk[:1200])
````

## Diagnostic: chunk quality check


````python
# ── Diagnostic: check chunk quality before calling Azure ─────────
def chunk_quality_report(section_chunks: dict) -> dict:
    report = {}
    for section, chunk in section_chunks.items():
        words = chunk.split()
        long_unspaced = re.findall(r"\b[A-Za-z]{22,}\b", chunk)
        repeated_artifacts = re.findall(r"([A-Za-z])\1{2,}", chunk)
        report[section] = {
            "word_count": len(words),
            "starts_with": chunk[:120],
            "long_unspaced_words_count": len(long_unspaced),
            "long_unspaced_words_sample": long_unspaced[:10],
            "repeated_letter_artifact_count": len(repeated_artifacts),
            "looks_like_table_of_contents": is_table_of_contents_page(chunk),
            "looks_garbled_or_concatenated": looks_garbled_or_concatenated(chunk),
        }
    return report

quality = chunk_quality_report(section_chunks)
print(json.dumps(quality, indent=2, ensure_ascii=False))
````

````python
# ── Step 5: Azure OpenAI API helper ─────────────────────────
def parse_json_response(raw: str) -> dict:
    '''
    Strip accidental markdown fences and parse JSON.
    '''
    raw = raw.strip()
    raw = re.sub(r"^```json\s*", "", raw)
    raw = re.sub(r"^```\s*", "", raw)
    raw = re.sub(r"\s*```$", "", raw)
    return json.loads(raw)


def get_azure_openai_client() -> AzureOpenAI:
    '''
    Build Azure OpenAI client from environment variables or manual config cell values.
    '''
    if AzureOpenAI is None:
        raise ImportError("openai package is not installed. Run the install cell first.")

    missing = []
    if not AZURE_OPENAI_API_KEY:
        missing.append("AZURE_OPENAI_API_KEY")
    if not AZURE_OPENAI_ENDPOINT:
        missing.append("AZURE_OPENAI_ENDPOINT")
    if not AZURE_OPENAI_DEPLOYMENT:
        missing.append("AZURE_OPENAI_DEPLOYMENT or AZURE_OPENAI_CHAT_DEPLOYMENT")

    if missing:
        raise EnvironmentError(
            "Missing Azure OpenAI settings: " + ", ".join(missing) +
            ". Set them in your .env file or in the configuration cell."
        )

    return AzureOpenAI(
        api_key=AZURE_OPENAI_API_KEY,
        azure_endpoint=AZURE_OPENAI_ENDPOINT,
        api_version=AZURE_OPENAI_API_VERSION,
    )


def azure_chat_json(
    client: AzureOpenAI,
    system_prompt: str,
    user_prompt: str,
    max_tokens: int = 2000,
    temperature: float = 0,
) -> dict:
    '''
    Calls Azure OpenAI and returns parsed JSON.
    Uses JSON response format when supported; falls back to normal text JSON if needed.
    '''
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt},
    ]

    try:
        response = client.chat.completions.create(
            model=AZURE_OPENAI_DEPLOYMENT,
            messages=messages,
            max_tokens=max_tokens,
            temperature=temperature,
            response_format={"type": "json_object"},
        )
    except Exception as first_error:
        # Some Azure deployments/API versions may not support response_format.
        log.warning(f"Retrying without response_format because first call failed: {first_error}")
        response = client.chat.completions.create(
            model=AZURE_OPENAI_DEPLOYMENT,
            messages=messages,
            max_tokens=max_tokens,
            temperature=temperature,
        )

    raw = response.choices[0].message.content.strip()
    return parse_json_response(raw)


def analyze_section_with_azure(
    client: AzureOpenAI,
    section_name: str,
    text_chunk: str,
    debug: bool = False
) -> Optional[dict]:
    '''
    Send a section chunk to Azure OpenAI for style analysis.
    Returns parsed JSON dict or None on failure.
    '''
    if not text_chunk.strip():
        log.warning(f"Empty chunk for section '{section_name}' — skipping")
        return None

    prompt_template = SECTION_PROMPTS.get(section_name, SECTION_PROMPTS["general"])
    prompt = prompt_template.format(text=text_chunk)

    if debug:
        log.info(f"Sending {len(text_chunk.split())} words to Azure OpenAI for '{section_name}'")

    try:
        parsed = azure_chat_json(
            client=client,
            system_prompt=(
                "You are a document style analyst. "
                "You always respond with valid JSON only — no markdown, no preamble, no explanation. "
                "If you cannot extract a field, use null. Never invent data not present in the text."
            ),
            user_prompt=prompt,
            max_tokens=2000,
            temperature=0,
        )

        log.info(f"  ✓ '{section_name}' analysis complete")
        return parsed

    except json.JSONDecodeError as e:
        log.error(f"JSON parse error for '{section_name}': {e}")
        return None
    except Exception as e:
        log.error(f"Azure OpenAI API error for '{section_name}': {e}")
        return None


print("Azure OpenAI helper functions ready")
````

````python
# ── Step 6: Synthesis prompt ────────────────────────────────
SYNTHESIS_PROMPT = """
You are a document style consultant. You have analyzed sections of an IFRS S1/S2 sustainability report.

Here are the per-section style analyses:
{section_analyses}

Synthesize these into a unified style guide for use by AI agents generating new report sections.

Return ONLY valid JSON — no preamble, no markdown fences:

{
  "report_identity": {
    "organization_name": "how the bank refers to itself in text",
    "reporting_year": "year or period reference style used",
    "document_type": "type of document style (formal corporate report, etc.)"
  },
  "universal_style_rules": [
    "8-10 rules that apply across ALL sections, written as actionable instructions for an AI writer"
  ],
  "tone_profile": {
    "primary_tone": "single best descriptor",
    "secondary_tones": ["2-3 secondary descriptors"],
    "what_to_avoid": ["3-5 tone/style anti-patterns NOT present in this report"]
  },
  "section_specific_instructions": {
    "governance": "2-3 sentence writing instruction for governance section agent",
    "strategy": "2-3 sentence writing instruction for strategy section agent",
    "risk_management": "2-3 sentence writing instruction for risk management section agent",
    "metrics_and_targets": "2-3 sentence writing instruction for metrics section agent"
  },
  "common_opening_patterns": ["5 ways sections or paragraphs typically begin"],
  "common_closing_patterns": ["3 ways sections or paragraphs typically end"],
  "master_phrase_bank": ["20-25 exact phrases or sentence starters from the report to reuse in generation"],
  "formatting_conventions": {
    "heading_style": "description of heading formatting",
    "table_style": "description of how tables are presented",
    "list_style": "description of bullet/numbered list usage",
    "number_formatting": "how numbers and percentages are written"
  }
}
"""

print("SYNTHESIS_PROMPT ready")
````

````python
# ── Step 6: Synthesize unified style guide ──────────────────
def synthesize_style_guide(client: AzureOpenAI, section_analyses: dict) -> Optional[dict]:
    '''
    Call Azure OpenAI to synthesize all section analyses into a unified style guide.
    '''
    log.info("Synthesizing unified style guide...")

    valid_analyses = {k: v for k, v in section_analyses.items() if v is not None}

    if not valid_analyses:
        log.error("No valid section analyses to synthesize")
        return None

    prompt = SYNTHESIS_PROMPT.format(
        section_analyses=json.dumps(valid_analyses, indent=2, ensure_ascii=False)
    )

    try:
        parsed = azure_chat_json(
            client=client,
            system_prompt=(
                "You are a document style consultant. "
                "Respond with valid JSON only. No markdown, no preamble."
            ),
            user_prompt=prompt,
            max_tokens=3000,
            temperature=0,
        )

        log.info("  ✓ Synthesis complete")
        return parsed

    except json.JSONDecodeError as e:
        log.error(f"Synthesis JSON parse error: {e}")
        return None
    except Exception as e:
        log.error(f"Synthesis Azure OpenAI API error: {e}")
        return None


style_guide = synthesize_style_guide(client, section_analyses)

print(json.dumps(style_guide, indent=2, ensure_ascii=False) if style_guide else "No style guide generated")
````

````python
# ── Step 7: Build LangGraph agent prompts ───────────────────
def build_agent_prompts(style_guide: dict) -> dict[str, str]:
    '''
    Build system prompts for each LangGraph section agent using extracted style patterns.
    '''
    if not style_guide:
        return {}

    universal_rules = "\n".join(
        f"- {rule}"
        for rule in style_guide.get("universal_style_rules", [])
    )

    phrase_bank = "\n".join(
        f'- "{phrase}"'
        for phrase in style_guide.get("master_phrase_bank", [])
    )

    org_name = style_guide.get("report_identity", {}).get("organization_name", "the Group")
    primary_tone = style_guide.get("tone_profile", {}).get("primary_tone", "formal")

    prompts = {}

    for section in ["governance", "strategy", "risk_management", "metrics_and_targets"]:
        section_instruction = (
            style_guide
            .get("section_specific_instructions", {})
            .get(section, "Write in a formal, professional tone.")
        )

        avoid = "\n".join(
            f"- {item}"
            for item in style_guide.get("tone_profile", {}).get("what_to_avoid", [])
        )

        prompts[section] = f'''You are an IFRS S1/S2 sustainability reporting specialist writing for {org_name}.

SECTION: {section.upper().replace("_", " ")}

STYLE INSTRUCTION:
{section_instruction}

UNIVERSAL STYLE RULES:
{universal_rules}

TONE: {primary_tone}

AVOID:
{avoid}

PHRASE BANK:
{phrase_bank}

TASK:
Given the JSON input data for the {section.replace("_", " ")} section, write the complete disclosure narrative.

Rules:
- Do not invent data not present in the input JSON.
- If a required disclosure is missing, state the evidence boundary in report-style wording.
- Match the style patterns above.
- Output the section text only.
'''

    return prompts


agent_prompts = build_agent_prompts(style_guide)

print("Agent prompts generated:", len(agent_prompts))
for section, prompt in agent_prompts.items():
    print("\n" + "="*80)
    print(section.upper())
    print("="*80)
    print(prompt[:1200])
````
